# Importing Libraries

```python
import time
import numpy as np
import os
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from xgboost import XGBClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
```

# Define the dataset folders

```python
dataset_folders = [
    "../Dataset/Train Test Dataset - Feature 5",
    "../Dataset/Train Test Dataset - Feature 8",
    "../Dataset/Train Test Dataset - Feature 16",
    "../Dataset/Train Test Dataset - Feature 22",
    "../Dataset/Train Test Dataset - Feature 30"
]
```

# Machine Learning Algorithms

```python
# Define classifiers
classifiers = {
    'KNN': KNeighborsClassifier(),
    'Random Forest': RandomForestClassifier(),
    'XGBoost': XGBClassifier(tree_method='gpu_hist', n_estimators=1000),
    'Decision Tree': DecisionTreeClassifier(),
    'Extra Trees': ExtraTreesClassifier()
}

# Loop through each dataset folder
for folder in dataset_folders:
    print(f"\nDataset: {folder}\n{'=' * 50}")

    # Load dataset
    X_train = np.load(os.path.join(folder, 'X_train.npy'))
    X_test = np.load(os.path.join(folder, 'X_test.npy'))
    y_train = np.load(os.path.join(folder, 'y_train.npy'))
    y_test = np.load(os.path.join(folder, 'y_test.npy'))

    # Loop through each classifier
    for clf_name, clf in classifiers.items():
        print(f"\nClassifier: {clf_name}\n{'-' * 30}")
        
        # Measure training time
        start_time = time.time()
        clf.fit(X_train, y_train)
        training_time = time.time() - start_time

        # Predict on training and testing data
        start_time = time.time()
        y_pred_train = clf.predict(X_train)
        y_pred_test = clf.predict(X_test)
        testing_time = time.time() - start_time

        # Calculate metrics
        training_metrics = [
            accuracy_score(y_train, y_pred_train),
            precision_score(y_train, y_pred_train),
            recall_score(y_train, y_pred_train),
            f1_score(y_train, y_pred_train),
            roc_auc_score(y_train, y_pred_train)
        ]

        testing_metrics = [
            accuracy_score(y_test, y_pred_test),
            precision_score(y_test, y_pred_test),
            recall_score(y_test, y_pred_test),
            f1_score(y_test, y_pred_test),
            roc_auc_score(y_test, y_pred_test)
        ]

        # Print metrics
        print("\nTraining Metrics:")
        print(f"Accuracy: {training_metrics[0]:.2f}, Precision: {training_metrics[1]:.2f}, "
              f"Recall: {training_metrics[2]:.2f}, F1-score: {training_metrics[3]:.2f}, "
              f"AUC-ROC: {training_metrics[4]:.2f}")
        print(f"Training Time (s): {training_time:.4f}")

        print("\nTesting Metrics:")
        print(f"Accuracy: {testing_metrics[0]:.2f}, Precision: {testing_metrics[1]:.2f}, "
              f"Recall: {testing_metrics[2]:.2f}, F1-score: {testing_metrics[3]:.2f}, "
              f"AUC-ROC: {testing_metrics[4]:.2f}")
        print(f"Testing Time (s): {testing_time:.4f}")
```
